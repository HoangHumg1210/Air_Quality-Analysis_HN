```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

y = pd.Series([10, 15, 20, 40, 100, 150])  # giả sử PM2.5
plt.plot(y, label='Gốc')
plt.legend(); plt.show()
```

```python
from scipy import stats
import pandas as pd

x = pd.Series([10, 12, 15, 20, 200, 300])

# Thực hiện Box-Cox
y, lam = stats.boxcox(x)

print("Lambda:", lam)
print("Dữ liệu sau Boxcox:", y)
```

```python
# giả sử dự báo log là:
yhat_log = pd.Series([2.5, 3.0, 4.0, 4.5])
# hoàn nguyên về thang gốc
yhat = np.expm1(yhat_log)
print(yhat)
```
