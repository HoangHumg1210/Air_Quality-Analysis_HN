```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

y = pd.Series([10, 15, 20, 40, 100, 150])  # giả sử PM2.5
plt.plot(y, label='Gốc')
plt.legend(); plt.show()
```

```python
y_log = np.log1p(y)
plt.plot(y_log, label='Log1p')
plt.legend(); plt.show()
```

```python
# giả sử dự báo log là:
yhat_log = pd.Series([2.5, 3.0, 4.0, 4.5])
# hoàn nguyên về thang gốc
yhat = np.expm1(yhat_log)
print(yhat)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 11)
y = np.array([5, 10, 12, 15, 20, 35, 80, 150, 200, 300])

plt.plot(x, y, 'bo-')
plt.title("PM2.5 gốc (rất lệch phải)")
plt.show()
```

```python
y_log = np.log1p(y)
plt.plot(x, y_log, 'ro-')
plt.title("PM2.5 sau khi log1p (đều hơn, dễ học hơn)")
plt.show()
```
