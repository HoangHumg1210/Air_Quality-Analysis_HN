```python
import pandas as pd


# Load 2 file
df_train = pd.read_csv("E:\Document\PROJECT_1\data\hanoi-aqi-weather-data-TEST.csv")
df_test  = pd.read_csv("hanoi-aqi-weather-data-TEST_1.csv")   # hoặc TEST_1.csv

# Chuyển về datetime
df_train["Local Time"] = pd.to_datetime(df_train["Local Time"])
df_test["Local Time"]  = pd.to_datetime(df_test["Local Time"])

# Gộp ngang theo dòng (append)
df_full = pd.concat([df_train, df_test], ignore_index=True)

# Sắp xếp theo thời gian
df_full = df_full.sort_values("Local Time")

# Đặt index
df_full = df_full.set_index("Local Time")


```
